```python
import openmeteo_requests
import time
import pandas as pd
import requests_cache
from retry_requests import retry
import calendar

# ---- Setup API client with caching + retries ----
cache_session = requests_cache.CachedSession('.cache', expire_after=-1)
retry_session = retry(cache_session, retries=5, backoff_factor=0.2)
openmeteo = openmeteo_requests.Client(session=retry_session)

# ---- Load your CSV ----
df = pd.read_csv("./data-2014-2021-hopefully-final.csv")

# ---- Normalize weekday column ----
df["day_of_week"] = df["day_of_week"].astype(str).str.strip().str.lower()

# Map weekday strings (English + Finnish + abbreviations)
weekday_map = {
    "monday": 0, "tuesday": 1, "wednesday": 2,
    "thursday": 3, "friday": 4, "saturday": 5, "sunday": 6,
    "maanantai": 0, "tiistai": 1, "keskiviikko": 2,
    "torstai": 3, "perjantai": 4, "lauantai": 5, "sunnuntai": 6,
    "mon": 0, "tue": 1, "wed": 2, "thu": 3, "fri": 4, "sat": 5, "sun": 6
}

# ---- Helper: dates of all given weekdays in month ----
def dates_for_weekday(year, month, weekday_name):
    if weekday_name not in weekday_map:
        return []
    start_date = f"{int(year)}-{int(month):02d}-01"
    last_day = calendar.monthrange(int(year), int(month))[1]
    end_date = f"{int(year)}-{int(month):02d}-{last_day}"
    all_dates = pd.date_range(start=start_date, end=end_date)
    weekday_num = weekday_map[weekday_name]
    return all_dates[all_dates.weekday == weekday_num].strftime("%Y-%m-%d").tolist()

# ---- Chunk Config ----
CHUNK_SIZE = 5000
#TOTAL_ROWS = len(df)
START_ROW = 95000   # <-- change this if resuming from a later point
TOTAL_ROWS = START_ROW+5000
totalapicalls = 0
# ---- Track skipped rows ----
skipped_rows = []

# ---- Process in Chunks ----
for start in range(START_ROW, TOTAL_ROWS, CHUNK_SIZE):
    end = min(start + CHUNK_SIZE, TOTAL_ROWS)
    print(f"\nProcessing rows {start} to {end-1}...")

    df_chunk = df.iloc[start:end].copy()
    snowfall_cols = [f"snowfall_{i}" for i in range(1, 6)]
    for col in snowfall_cols:
        df_chunk[col] = None

    url = "https://archive-api.open-meteo.com/v1/archive"

    # --- Step 1: build unique combos in this chunk ---
    combos = df_chunk[["lat", "lon", "year", "month"]].drop_duplicates()
    combo_results = {}

    for _, row_combo in combos.iterrows():
        lat, lon, year, month = row_combo["lat"], row_combo["lon"], int(row_combo["year"]), int(row_combo["month"])

        if pd.isna(lat) or pd.isna(lon):
            combo_results[(lat, lon, year, month)] = None
            continue

        start_date = f"{year}-{month:02d}-01"
        last_day = calendar.monthrange(year, month)[1]
        end_date = f"{year}-{month:02d}-{last_day}"

        params = {
            "latitude": lat,
            "longitude": lon,
            "start_date": start_date,
            "end_date": end_date,
            "daily": ["snowfall_sum"],
            "timezone": "GMT+2"
        }

        try:
            responses = openmeteo.weather_api(url, params=params)
        except Exception as e:
            if "limit" in str(e).lower():
                print("Rate limit hit. Waiting 60s...")
                time.sleep(60)
                try:
                    responses = openmeteo.weather_api(url, params=params)
                    totalapicalls += 1
                except Exception as e2:
                    combo_results[(lat, lon, year, month)] = None
                    print(f"API failed again for {lat},{lon} {year}-{month}: {e2}")
                    continue
            else:
                combo_results[(lat, lon, year, month)] = None
                print(f"API error for {lat},{lon} {year}-{month}: {e}")
                continue

        response = responses[0]
        daily = response.Daily()
        snowfall = daily.Variables(0).ValuesAsNumpy()

        daily_data = pd.DataFrame({
            "date": pd.date_range(
                start=pd.to_datetime(daily.Time(), unit="s", utc=True),
                end=pd.to_datetime(daily.TimeEnd(), unit="s", utc=True),
                freq=pd.Timedelta(seconds=daily.Interval()),
                inclusive="left"
            ).strftime("%Y-%m-%d"),
            "snowfall_sum": snowfall
        })

        combo_results[(lat, lon, year, month)] = daily_data

    # --- Step 2: fill in snowfall data for each row ---
    for i, row in df_chunk.iterrows():
        year, month, weekday_name = int(row["year"]), int(row["month"]), row["day_of_week"]
        lat, lon = row["lat"], row["lon"]
        key = (lat, lon, year, month)

        if key not in combo_results or combo_results[key] is None:
            skipped_rows.append((row["id"], f"no API data for combo {key}"))
            continue

        weekday_dates = dates_for_weekday(year, month, weekday_name)
        if not weekday_dates:
            skipped_rows.append((row["id"], f"invalid weekday '{weekday_name}'"))
            continue

        daily_data = combo_results[key]
        selected = daily_data[daily_data["date"].isin(weekday_dates)]["snowfall_sum"].tolist()

        if not selected:
            skipped_rows.append((row["id"], f"no snowfall data for weekday {weekday_name}"))
            continue

        for j, val in enumerate(selected[:5]):
            df_chunk.at[i, f"snowfall_{j+1}"] = val

    # ---- Save this chunk ----
    output_file = f"output_{start}_{end-1}.csv"
    df_chunk.to_csv(output_file, index=False)
    print(f"Saved {output_file}")
    print(f"{totalapicalls} API calls made")

# ---- Save skipped rows log ----
if skipped_rows:
    skipped_df = pd.DataFrame(skipped_rows, columns=["id", "reason"])
    skipped_df.to_csv("skipped_rows_log.csv", index=False)
    print(f"Logged {len(skipped_rows)} skipped rows to skipped_rows_log.csv")
```
